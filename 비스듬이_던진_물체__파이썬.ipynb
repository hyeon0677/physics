```python
import math

g = 9.8
angle_deg = 37.0
PI = 3.141592
v0 = 100.0
y0 = 140.0

angle_rad = angle_deg * PI / 180.0
v0x = v0 * math.cos(angle_rad)
v0y = v0 * math.sin(angle_rad)

t = (v0y + math.sqrt(v0y * v0y + 2 * g * y0)) / g
print(f"지면에 닿기 까지 걸린 시간 t = {t:.2f}초")

x = v0x * t
print(f"절벽 아래로부터 수평거리 x = {x:.2f}m")

vx = v0x
vy = v0y - g * t
print(f"수평 속도 vx = {vx:.2f}m/s")
print(f"수직 속도 vy = {vy:.2f}m/s")

V = math.sqrt(vx * vx + vy * vy)
print(f"속도의 크기 V = {V:.2f}m/s")

tan_theta = abs(vy / vx)
angle_rad_2 = math.atan(tan_theta)
angle_deg_2 = angle_rad_2 * 180.0 / PI
print(f"속도 벡터가 수평선과 이루는 각 = {angle_deg_2:.2f}도")
```
