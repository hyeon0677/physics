```python
#include <stdio.h>

int main()
{
    // 변수 선언
    double x1 = 0.0;
    double y1 = 0.0;
    double z1 = 0.0;

    double x2 = 0.0;
    double y2 = 0.0;
    double z2 = 0.0;

    double x = 0.0;
    double y = 0.0;
    double z = 0.0;

    double dot_product = 0.0;
    double cross_product = 0.0;

    // 첫 번째 벡터 x, y, z 값을 입력
    printf("x1 좌표를 실수로 입력하세요 : ");
    scanf_s("%lf", &x1);
    printf("y1 좌표를 실수로 입력하세요 : ");
    scanf_s("%lf", &y1);
    printf("z1 좌표를 실수로 입력하세요 : ");
    scanf_s("%lf", &z1);

    // 두 번째 벡터 x, y, z 값 입력
    printf("x2 좌표를 실수로 입력하세요 : ");
    scanf_s("%lf", &x2);
    printf("y2 좌표를 실수로 입력하세요 : ");
    scanf_s("%lf", &y2);
    printf("z2 좌표를 실수로 입력하세요 : ");
    scanf_s("%lf", &z2);

    // 두 벡터의 합 계산
    x = (x1 + x2);
    y = (y1 + y2);
    z = (z1 + z2);

    // 내적 값 계산
    dot_product = (x1 * y1 + x2 * y2 + z1 * z2);

    // 외적 값 계산
    cross_product = (y1 * z2 - z1 * y2 + z1 * x2 - x1 * z2 + x1 * y2 - y1 * x2);

    // 결과 출력
    printf("두 벡터의 합 : %.1lf , %.1lf , %.1lf \n", x, y, z);
    printf("내적 값 계산 : %.1lf \n", dot_product);
    printf("외적 값 계산 : %.1lf \n", cross_product);

    return 0;
}
```
