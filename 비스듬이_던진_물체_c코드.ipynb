```python
#include <stdio.h>
#include <math.h>

int main(void) {
	const double g = 9.8;
	const double angle_deg = 37.0;
	const double PI = 3.141592;
	const double v0 = 100.0;
	const double y0 = 140.0;

	double angle_rad = angle_deg * PI / 180.0;

	double v0x = v0 * cos(angle_rad);
	double v0y = v0 * sin(angle_rad);

	// 포사체가 지면에 충돌할 때 까지 걸린 시간
	double t = (v0y + sqrt(v0y * v0y + 2 * g * y0)) / g;
		printf("지면에 닿기 까지 걸린 시간 t = %.2f초\n", t);

	// 절벽 아래로부터 수평거리 x (점 P까지의)
	double x = v0x * t;
		printf("절벽 아래로부터 수평거리 x : %.2fm\n", x);

	// 점 P에 도달하기 까지 걸린 수평 / 수직 속도
	double vx = v0x;
	double vy = v0y - g * t;
		printf("수평 속도 : %.2fm/s\n", vx);
		printf("수직 속도 : %.2fm/s\n", vy);

	// 속도의 크기
		double V = sqrt((vx * vx) + (vy * vy));
		printf("속도의 크기 = %.2fm/s\n", V);

	// 속도 벡터가 수평선과 이루는 각
		double tan_theta = fabs(vy / vx);
		double angle_rad_2 = atan(tan_theta);
		double angle_deg_2 = angle_rad_2 * 180.0 / PI;
		printf("속도 벡터가 수평선과 이루는 각 : %.2f도\n", angle_deg_2);

	return 0;
}
```
