```python
#include <stdio.h>
#include <math.h>

int main(void) {
    const double v0 = 30.0;
    const double angle_deg = 30.0;
    const double PI = 3.141592;
    const double g = 9.8;

    double angle_rad = angle_deg * PI / 180.0;

    // x, y 의 초기 속도
    double v0x = v0 * cos(angle_rad);
    double v0y = v0 * sin(angle_rad);

    printf("초기속도 v0x = %.1f m/s\n", v0x);
    printf("초기속도 v0y = %.1f m/s\n", v0y);

    // 최고점의 높이
    double H = (v0y * v0y) / (2 * g);
    printf("최고점 높이 = %.1f m\n", H);

    // 최고점 도달 시간
    double t = v0y / g;
    printf("최고점 도달 시간 = %.1f s\n", t);

    // 수평으로 이동한 최고 거리
    double total_time = 2 * t;
    double range = v0x * total_time;
    printf("수평 최고 거리 = %.1f m\n", range);

    // 지면 도달 시 속도 및 V 벡터 성분
    double vxf = v0x;
    double vyf = -v0y;
    double V = sqrt(vxf * vxf + vyf * vyf);
    printf("지면 도달 시 속도 v0x = %.1f m/s, v0y = %.1f m/s\n", vxf, vyf);
    printf("전체 속도 크기 = %.1f m/s\n", V);

    return 0;
}
```
