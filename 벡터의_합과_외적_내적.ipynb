```python
#include <stdio.h>

int main() {
	double x1, y1, x2, y2, x, y, dot_product, cross_product = 0.0;

	printf("x1 좌표를 실수로 입력하세요 : ");
	scanf_s("%lf", &x1);
	printf("y1 좌표를 실수로 입력하세요 : ");
	scanf_s("%lf", &y1);

	printf("x2 좌표를 실수로 입력하세요 : ");
	scanf_s("%lf", &x2);
	printf("y2 좌표를 실수로 입력하세요 : ");
	scanf_s("%lf", &y2);

	x = (x1 + x2);
	y = (y1 + y2);

	dot_product = (x1 * y1 + x2 * y2);
	cross_product = (x1 * y2 - x2 * y1);

	printf("두 벡터의 합 : [ %.1f , %.1f ]\n", x, y);
	printf("내적 값 계산 : [ %.1f * %.1f + %.1f * %.1f = %.1f]\n", x1, y1, x2, y2, dot_product);
	printf("외적 값 계산 : [ %.1f * %.1f - %.1f * %.1f = %.1f]", x1, y2, x2, y1, cross_product);
}
```
