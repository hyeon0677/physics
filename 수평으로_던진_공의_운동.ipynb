```python
import math

# 기본 값
v0 = 30
angle = 30
g = 9.8

# 라디안 변환
rad = math.radians(angle)

# 초기속도 x, y 방향
vx = v0 * math.cos(rad)
vy = v0 * math.sin(rad)
print("Vx =", round(vx, 2), "m/s,", "Vy =", round(vy, 2), "m/s")

# 최고 높이
h = (vy ** 2) / (2 * g)
print("최고 높이 =", round(h, 2), "m")

# 최고점 도달 시간
t = vy / g
print("최고점 도달 시간 =", round(t, 2), "s")

# 수평 총 이동 거리
range_x = vx * (2 * t)
print("수평 거리 =", round(range_x, 2), "m")

# 지면 도달 시 속도 성분
vy_f = -vy
v_t = math.sqrt(vx**2 + vy_f**2)
print("지면 도달시 Vx =", round(vx,2), "Vy =", round(vy_f,2), "속도 크기 =", round(v_t,2))
```
